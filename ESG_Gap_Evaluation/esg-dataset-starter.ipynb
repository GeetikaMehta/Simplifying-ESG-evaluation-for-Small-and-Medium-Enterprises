```python
import matplotlib.pyplot as plt
```

```python
value_to_count1 = "A"
value_to_count2 = "B"
count1 = df["environment_grade"].value_counts()[value_to_count1]
count2 = df["environment_grade"].value_counts()[value_to_count2]
```

```python
plt.bar([value_to_count1, value_to_count2], [count1, count2])
plt.xlabel('Environment Grade')
plt.ylabel('Occurrences')
plt.title('Occurrences of ' + value_to_count1 + " and " + value_to_count2)
plt.show()
```
